### Задание 1. Периметр квадрата
Дана сторона квадрата a. Найти его периметр P = 4 * a.


```python
a = float(input("Введите сторону квадрата a: "))
perimeter = 4.0 * a
print(f"Периметр квадрата P = {perimeter}")
```

### Задание 2. Площадь квадрата
Дана сторона квадрата a. Найти его площадь S = a^2.


```python
a = float(input("Введите сторону квадрата a: "))
area = a ** 2
print(f"Площадь квадрата S = {area}")
```

### Задание 3. Площадь и периметр прямоугольника
Даны стороны прямоугольника a и b. Найти его площадь S = a * b и периметр P = 2 * (a + b).


```python
a = float(input("Введите сторону прямоугольника a: "))
b = float(input("Введите сторону прямоугольника b: "))
area = a * b
perimeter = 2.0 * (a + b)
print(f"Площадь прямоугольника S = {area}")
print(f"Периметр прямоугольника P = {perimeter}")
```

### Задание 4. Длина окружности по диаметру
Дан диаметр окружности d. Найти ее длину L = pi * d. В качестве значения pi использовать 3.14.


```python
PI = 3.14
d = float(input("Введите диаметр окружности d: "))
length = PI * d
print(f"Длина окружности L = {length}")
```

### Задание 5. Объем и площадь поверхности куба
Дана длина ребра куба a. Найти объем куба V = a^3 и площадь его поверхности S = 6 * a^2.


```python
a = float(input("Введите длину ребра куба a: "))
volume = a ** 3
surface_area = 6.0 * (a ** 2)
print(f"Объем куба V = {volume}")
print(f"Площадь поверхности куба S = {surface_area}")
```

### Задание 6. Объем и площадь поверхности параллелепипеда
Даны длины ребер a, b, c прямоугольного параллелепипеда. Найти его объем V = a * b * c и площадь поверхности S = 2 * (a * b + b * c + a * c).


```python
a = float(input("Введите длину ребра a: "))
b = float(input("Введите длину ребра b: "))
c = float(input("Введите длину ребра c: "))
volume = a * b * c
surface_area = 2.0 * (a * b + b * c + a * c)
print(f"Объем параллелепипеда V = {volume}")
print(f"Площадь поверхности S = {surface_area}")
```

### Задание 7. Длина окружности и площадь круга
Найти длину окружности L и площадь круга S заданного радиуса R: L = 2 * pi * R, S = pi * R^2. В качестве значения pi использовать 3.14.


```python
PI = 3.14
r = float(input("Введите радиус круга R: "))
length = 2.0 * PI * r
area = PI * (r ** 2)
print(f"Длина окружности L = {length}")
print(f"Площадь круга S = {area}")
```

### Задание 8. Среднее арифметическое двух чисел
Даны два числа a и b. Найти их среднее арифметическое: (a + b) / 2.


```python
a = float(input("Введите первое число a: "))
b = float(input("Введите второе число b: "))
arithmetic_mean = (a + b) / 2.0
print(f"Среднее арифметическое = {arithmetic_mean}")
```

### Задание 9. Среднее геометрическое двух неотрицательных чисел
Даны два неотрицательных числа a и b. Найти их среднее геометрическое: sqrt(a * b).


```python
a = float(input("Введите первое число a: "))
b = float(input("Введите второе число b: "))
if a < 0 or b < 0:
    print("Ошибка: числа должны быть неотрицательными!")
else:
    geometric_mean = (a * b) ** 0.5
    print(f"Среднее геометрическое = {geometric_mean}")
```

### Задание 10. Операции над квадратами ненулевых чисел
Даны два ненулевых числа. Найти сумму, разность, произведение и частное их квадратов.


```python
a = float(input("Введите первое число a: "))
b = float(input("Введите второе число b: "))
if a == 0 or b == 0:
    print("Ошибка: числа должны быть ненулевыми!")
else:
    sq_a = a ** 2
    sq_b = b ** 2
    print(f"Сумма квадратов = {sq_a + sq_b}")
    print(f"Разность квадратов = {sq_a - sq_b}")
    print(f"Произведение квадратов = {sq_a * sq_b}")
    print(f"Частное квадратов = {sq_a / sq_b}")
```

## Часть 2. Условный оператор
*Изучение условных инструкций if, elif, else и операций сравнения.*


### Задание 1. Модификация положительного числа
Дано целое число. Если оно является положительным, то прибавить к нему 1; в противном случае не изменять его. Вывести полученное число.


```python
n = int(input("Введите целое число: "))
if n > 0:
    n += 1
print(f"Полученное число: {n}")
```

### Задание 2. Двухвариантное изменение числа
Дано целое число. Если оно является положительным, то прибавить к нему 1; в противном случае вычесть из него 2. Вывести полученное число.


```python
n = int(input("Введите целое число: "))
if n > 0:
    n += 1
else:
    n -= 2
print(f"Полученное число: {n}")
```

### Задание 3. Трёхвариантное ветвление
Дано целое число. Если оно является положительным, то прибавить к нему 1; если отрицательным, то вычесть из него 2; если нулевым, то заменить его на 10. Вывести полученное число.


```python
n = int(input("Введите целое число: "))
if n > 0:
    n += 1
elif n < 0:
    n -= 2
else:
    n = 10
print(f"Полученное число: {n}")
```

### Задание 4. Подсчет количества положительных чисел
Даны три целых числа. Найти количество положительных чисел в исходном наборе.


```python
a = int(input("Введите первое целое число: "))
b = int(input("Введите второе целое число: "))
c = int(input("Введите третье целое число: "))
positive_count = 0
if a > 0: positive_count += 1
if b > 0: positive_count += 1
if c > 0: positive_count += 1
print(f"Количество положительных чисел: {positive_count}")
```

### Задание 5. Подсчет количества положительных и отрицательных чисел
Даны три целых числа. Найти количество положительных и количество отрицательных чисел в исходном наборе.


```python
a = int(input("Введите первое целое число: "))
b = int(input("Введите второе целое число: "))
c = int(input("Введите третье целое число: "))
positive_count = 0
negative_count = 0
for num in [a, b, c]:
    if num > 0:
        positive_count += 1
    elif num < 0:
        negative_count += 1
print(f"Количество положительных чисел: {positive_count}")
print(f"Количество отрицательных чисел: {negative_count}")
```
